## Єдине порівняння: усі 4 файли × усі бібліотеки

Однаковий код на всіх файлах з `docs/відпустка/`, щоб порівняння було чесним (не залежало від того, як саме готувався кожен файл окремо вище).

```python
# Підготовка всіх 4 файлів однаково
import pillow_avif
from pathlib import Path
from PIL import Image

BASE = Path.cwd().parent
SRC = BASE / "docs" / "відпустка"

FILES = {
    "raport-optimized (шаблон, чистий скан)": "raport-optimized.avif",
    "довідка ВЛК (фото паперу)": "__2023-10-17__121332.png",
    "public (порожній бланк)": "public.avif",
    "1.webp (чистий скан)": "1.webp",
}

CMP_IMAGES = {}
for label, fname in FILES.items():
    out_name = "cmp_" + Path(fname).stem + ".png"
    Image.open(SRC / fname).convert("RGB").save(out_name)
    CMP_IMAGES[label] = out_name

CMP_IMAGES
```

### Tesseract — усі 4 файли

```python
import pytesseract
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
from PIL import Image

for label, path in CMP_IMAGES.items():
    print(f"===== TESSERACT: {label} =====")
    text = pytesseract.image_to_string(Image.open(path), lang="ukr")
    print(text)
    print()
```

### EasyOCR — усі 4 файли

```python
import easyocr
reader = easyocr.Reader(["uk", "en"])

for label, path in CMP_IMAGES.items():
    print(f"===== EASYOCR: {label} =====")
    result = reader.readtext(path, detail=0, paragraph=True)
    print("\n".join(result))
    print()
```

### PaddleOCR — усі 4 файли

```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(
    text_recognition_model_name="cyrillic_PP-OCRv3_mobile_rec",
    use_textline_orientation=True,
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    enable_mkldnn=False
)

for label, path in CMP_IMAGES.items():
    print(f"===== PADDLEOCR: {label} =====")
    result = ocr.predict(input=path)
    for res in result:
        res.print()
    print()
```

### Surya OCR

```python
from PIL import Image
from surya.inference import SuryaInferenceManager
from surya.recognition import RecognitionPredictor

manager = SuryaInferenceManager()  # автоматично підніме vllm або llama.cpp сервер
recognition_predictor = RecognitionPredictor(manager)

for label, path in CMP_IMAGES.items():
    print(f"===== SURYA: {label} =====")
    predictions = recognition_predictor([Image.open(path)])
    for block in predictions[0].blocks:
        print(block.html)
    print()
```

```python
import re, html
from PIL import Image
from surya.inference import SuryaInferenceManager
from surya.recognition import RecognitionPredictor

manager = SuryaInferenceManager()
recognition_predictor = RecognitionPredictor(manager)

image_path = "cmp___2023-10-17__121332.png"  # довідка ВЛК
predictions = recognition_predictor([Image.open(image_path)])

for block in predictions[0].blocks:
    plain = re.sub(r"<br\s*/?>", "\n", block.html)  # <br/> -> новий рядок
    plain = re.sub(r"<[^>]+>", "", plain)            # решту тегів прибрати
    plain = html.unescape(plain).strip()              # &#39; тощо -> звичайні символи
    print(plain)
```
